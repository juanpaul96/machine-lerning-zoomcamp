```python
#Question 6
# step 1
hp_median = df["horsepower"].median()

# step 2
hp_mode = df["horsepower"].mode()[0]
# step 3
df_filled = df.copy()
df_filled["horsepower"] = df_filled["horsepower"].fillna(hp_mode)
# step 4
new_hp_median = df_filled["horsepower"].median()

print(f"previous median is: {hp_median} and new median is {new_hp_median}")
```

```python
#Question 7
# step 1
asia_cars_df = df[df["origin"] == 'Asia']
# step 2
asia_cars_df = asia_cars_df[["vehicle_weight","model_year"]]
# step 3
top_7_asia_cars = asia_cars_df.head(7)
# step 4
X = top_7_asia_cars.values
# step 5
XTX = X.T @ X
# step 6
XTX_inv = np.linalg.inv(XTX)
# step 7
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
# step 8
w = (XTX_inv @ X.T) * y
# step 9
w.sum()
```
